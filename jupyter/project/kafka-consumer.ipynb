```python
import json

from pprint import PrettyPrinter

from kafka import KafkaConsumer  # pip install kafka-python

from utils.config import load_config
```

```python
config = load_config("config.yaml")
config
```

```python
# `PrettyPrinter` ułatwia czytanie złożonych struktur danych w konsoli.
# Dzięki temu łatwiej zobaczyć treść otrzymanych komunikatów.
pprint = PrettyPrinter()
```

```python
# Warto odczytywać adres brokera i nazwę topicu z konfiguracji,
# zamiast wpisywać je na sztywno do kodu.
KAFKA_SERVER = f"{config['kafka']['host']}:{config['kafka']['port']}"
KAFKA_TOPIC_NAME = config["kafka"]["topic"]
```

```python
# `auto_offset_reset` decyduje, od którego miejsca zaczynamy odczyt,
# gdy grupa konsumentów rozpoczyna pracę po raz pierwszy.
consumer = KafkaConsumer(
    KAFKA_TOPIC_NAME,
    group_id=config["kafka"]["group-id"],
    bootstrap_servers=KAFKA_SERVER,
    auto_offset_reset=config["kafka"]["auto-offset-reset"],
)
```

```python
# Pętla `for msg in consumer` pobiera kolejne komunikaty tak długo,
# aż program zostanie zatrzymany przez użytkownika.
for msg in consumer:
    print(f"Otrzymano komunikat: {msg.value.decode('utf-8')}")
    d = json.loads(msg.value.decode("utf-8"))
    print("Otrzymany komunikat jako słownik (ładnie sformatowany):")
    pprint(d)
```
