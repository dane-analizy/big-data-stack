```python
# KafkaProducer publikuje zdarzenia do konkretnego topicu.
# Konfiguracja jest odczytywana z pliku config.yaml, aby nie hardcodować adresu brokera.
import json

from kafka import KafkaProducer  # pip install kafka-python
from utils.config import load_config
```

```python
config = load_config("config.yaml")
config
```

```python
# Warto ustawić endpoint brokera i nazwę topicu w jednym miejscu,
# bo takie dane są potrzebne zarówno do wysyłki, jak i odczytu wiadomości.
KAFKA_SERVER = f"{config['kafka']['host']}:{config['kafka']['port']}"
KAFKA_TOPIC_NAME = config["kafka"]["topic"]
```

```python
# bootstrap_servers wskazuje, gdzie działa Kafka.
# acks="all" oznacza, że komunikat zostanie zapisany na wszystkich replikach,
# zanim producent uzna wysyłkę za pomyślną.
producer = KafkaProducer(
    bootstrap_servers=[KAFKA_SERVER],
    acks="all",
    value_serializer=lambda v: json.dumps(v).encode("utf-8"),
)
```

```python
worked = False
```

```python
# Kafka przechowuje bajty, więc dane z Python trzeba zamienić na JSON.
# `flush()` gwarantuje, że wszystkie wiadomości zostały wysłane do brokera.
for i in range(10):
    d = {"i": i, "stala": "stala wartosc", "j": 1000 - i}
    d_str = json.dumps(d)
    print(f"Wysyłm: {d_str}")
    future = producer.send(KAFKA_TOPIC_NAME, value=d_str.encode('utf-8'))
    producer.flush()
```
